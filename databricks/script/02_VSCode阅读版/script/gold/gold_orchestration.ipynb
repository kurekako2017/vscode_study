## Orchestration Logic

## 🔗 学习说明：Gold Notebook Orchestration

这里使用：

```python
dbutils.notebook.run(...)
```

按 `notebooks` 列表顺序运行子 Notebook。

它是 **Notebook 内部编排**；你现在创建的 Databricks Job/Workflow 则是更上层的 **Task 编排**。


````python
notebooks = [
    "./gold_dim_customers",
    "./gold_dim_products",
    "./gold_fact_sales"
]

for nb in notebooks:
    print(f"Running {nb}")
    dbutils.notebook.run(nb, timeout_seconds=0)
````
