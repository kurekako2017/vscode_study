# Create Lakehouse Schemas

## 🏗️ 学习说明：创建 Bronze / Silver / Gold Schema

这一步是在 `workspace` Catalog 下面创建 Medallion 三层：

```text
bronze → 原始/接近原始状态
silver → 清洗、标准化后的可信数据
gold   → 面向分析/业务的数据模型
```

`IF NOT EXISTS` 表示对象已经存在时不会因为重复创建而失败。


````python
%sql
CREATE SCHEMA IF NOT EXISTS bronze
COMMENT 'Bronze layer: raw ingested data';

CREATE SCHEMA IF NOT EXISTS silver
COMMENT 'Silver layer: cleaned and transformed data';

CREATE SCHEMA IF NOT EXISTS gold
COMMENT 'Gold layer: business-ready data';
````

#Create Volume


## 📁 学习说明：创建原始文件 Volume

Volume 用于保存 **文件**，这里主要是 CRM / ERP CSV。

```text
/Volumes/workspace/bronze/raw_sources/
├── source_crm/
└── source_erp/
```

**Volume ≠ Table**

- Volume：保存 CSV/JSON/图片等文件
- Table：供 SQL/PySpark 查询和加工的正式数据资产


````python
%sql
CREATE VOLUME IF NOT EXISTS workspace.bronze.raw_sources
COMMENT 'Volume for raw source files (CSV)';
````
